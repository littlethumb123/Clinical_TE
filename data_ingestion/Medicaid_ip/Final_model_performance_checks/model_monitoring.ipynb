```python
sql = """
SELECT
    f.* EXCEPT(asdb_member_key)
    , p.ip_riskscore
FROM
    `anbc-hcb-prod.cm_medicaid_hcb_prod.medicaid_enhanced_ip_features` AS f
LEFT JOIN
    `anbc-hcb-prod.cm_medicaid_hcb_prod.medicaid_enhanced_ip_predict` AS p
        ON f.asdb_member_key = p.asdb_member_key
"""
curr_data = client.query(sql).to_dataframe() 

curr_data.shape
```

```python
sql = """
SELECT 
    f.* EXCEPT(asdb_member_key)
    , p.y_pred_test AS ip_riskscore
FROM 
    `anbc-hcb-dev.cm_medicaid_hcb_dev.a534354_dag_check_features` AS f
LEFT JOIN
    `anbc-hcb-dev.cm_medicaid_hcb_dev.a534354_medicaid_IP_predict` AS p
        ON f.asdb_member_key = p.asdb_member_key
"""
ref_data = client.query(sql).to_dataframe() 

ref_data.shape
```
